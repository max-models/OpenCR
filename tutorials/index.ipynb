# Tutorials

Jupyter notebooks in this section are executed when the documentation is built. Download a notebook from its page to run it yourself.

- [Atomic hydrogen](/OpenCR/tutorials/01-atomic-hydrogen/): populations, dominant processes and effective coefficients.
- [Emission and spectra](/OpenCR/tutorials/02-emission-and-spectra/): lines, photon emissivity coefficients, line ratios and a synthetic spectrum.
- [Ionization balance](/OpenCR/tutorials/03-ionization-balance/): closed steady states, quasi-neutrality and export.
- [Custom networks](/OpenCR/tutorials/04-custom-networks/): building, checking and saving a network, and non-Maxwellian electrons.
- [Time dependence](/OpenCR/tutorials/05-time-dependence/): transients with a stiff integrator, constant and time-dependent conditions.
- [Radiation trapping](/OpenCR/tutorials/06-radiation-trapping/): escape factors, the self-consistent solve, the absorption length.
- [Batched scans](/OpenCR/tutorials/07-batched-scans/): the quasi-static reduction of a whole grid at once on cunumpy.
- [JAX backend](/OpenCR/tutorials/08-jax-backend/): jitted rates, a vmapped scan and automatic differentiation.
- [Molecular hydrogen](/OpenCR/tutorials/09-molecular-hydrogen/): the vibrational distribution, effective rates of H₂ and negative ions.
- [Sensitivities and inference](/OpenCR/tutorials/10-sensitivities-and-inference/): sensitivities to the data, uncertainty propagation, $T_e$ and $n_e$ from line ratios.
- [Surrogates](/OpenCR/tutorials/11-surrogates/): a dataset with provenance, a baseline surrogate, held-out errors and the out-of-domain flag.

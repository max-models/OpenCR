# Tutorials

Jupyter notebooks in this section are executed when the documentation is built. Download a notebook from its page to run it yourself.

- [Atomic hydrogen](/opencr/tutorials/01-atomic-hydrogen/): populations, dominant processes and effective coefficients.
- [Emission and spectra](/opencr/tutorials/02-emission-and-spectra/): lines, photon emissivity coefficients, line ratios and a synthetic spectrum.
- [Ionization balance](/opencr/tutorials/03-ionization-balance/): closed steady states, quasi-neutrality and export.
- [Custom networks](/opencr/tutorials/04-custom-networks/): building, checking and saving a network, and non-Maxwellian electrons.
